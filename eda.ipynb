# Exploratory Data Analysis

```python
import numpy as np
import pandas as pd
import seaborn as sns
from matplotlib import pyplot as plt

import sklearn as sk
```

```python
X_train = pd.read_csv('data/X_train.csv',index_col='ROW_ID')
X_test = pd.read_csv('data/X_test.csv',index_col='ROW_ID')

y_train = pd.read_csv('data/y_train.csv',index_col='ROW_ID')
```

```python
print(X_train)
```

```python
# shapes dtype etc
print("X_train", X_train.shape, "X_test", X_test.shape, "y_train", y_train.shape)
print("index aligned:", X_train.index.equals(y_train.index))
print("\nNaN fraction per column (train, nonzero only):")
na = X_train.isna().mean()
print(na[na > 0].sort_values(ascending=False).round(4).to_string())
print("\nNaN fraction per column (test, nonzero only):")
na = X_test.isna().mean()
print(na[na > 0].sort_values(ascending=False).round(4).to_string())
print("\nrows with any NaN: train %.4f, test %.4f"
      % (X_train.isna().any(axis=1).mean(), X_test.isna().any(axis=1).mean()))
```
